```python
import pandas as pd
import numpy as np
import seaborn as sns

# Load the Titanic dataset
df = sns.load_dataset('titanic')

print(df.head())
print(df.shape)
```

#check missing values


```python
print(df.isnull().sum())
```

#handling missing values

```python
# Fill missing age values with the median
df['age'] = df['age'].fillna(df['age'].median())

# Fill missing embarkation values with the mode
df['embarked'] = df['embarked'].fillna(
    df['embarked'].mode()[0]
)

# Fill missing deck values with Unknown
df['deck'] = df['deck'].cat.add_categories(
    ['Unknown']
).fillna('Unknown')

print(df.isnull().sum())
```

#remove duplicate records

```python
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

Correct data types


```python
# Convert survived into an integer column
df['survived'] = df['survived'].astype(int)

# Convert passenger class into a category
df['pclass'] = df['pclass'].astype('category')

print(df.dtypes)
```

Detect and handle outliers

```python
# Find the lower and upper limits for age
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[
    (df['age'] < lower) | (df['age'] > upper)
]

print("Number of age outliers:", len(outliers))

# Remove age outliers
df = df[
    (df['age'] >= lower) & (df['age'] <= upper)
]

print("Shape after outlier removal:", df.shape)
```

check the cleaned dataset




```python
print("Final dataset shape:", df.shape)

print("Missing values:")
print(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())

print(df.head())
```
