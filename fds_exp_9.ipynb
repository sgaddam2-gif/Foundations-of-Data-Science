Take the IRIS dataset. Plot various visualization plots

```python

import pandas as pd

# 1. Create DataFrames
df1 = pd.DataFrame({
    'Student_ID': [1, 2, 3],
    'Name': ['Ravi', 'Sita', 'Arun'],
    'Department': ['CSE', 'ECE', 'CSE'],
    'Marks': [85, 90, 78]
})

df2 = pd.DataFrame({
    'Student_ID': [4, 5],
    'Name': ['Priya', 'Rahul'],
    'Department': ['ECE', 'CSE'],
    'Marks': [88, 92]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# 2. Hierarchical Indexing
hierarchical = df1.set_index(['Department', 'Student_ID'])
print("\nHierarchical Indexing:")
print(hierarchical)

# 3. Combining DataFrames using concat()
combined = pd.concat([df1, df2], ignore_index=True)
print("\nCombined DataFrames:")
print(combined)

# 4. Merging DataFrames
details = pd.DataFrame({
    'Student_ID': [1, 2, 3, 4, 5],
    'City': ['Delhi', 'Mumbai', 'Chennai',
             'Hyderabad', 'Bangalore']
})

merged = pd.merge(combined, details, on='Student_ID')
print("\nMerged DataFrame:")
print(merged)

# 5. Reshaping using melt()
reshaped = pd.melt(
    combined,
    id_vars=['Student_ID', 'Name'],
    value_vars=['Department', 'Marks'],
    var_name='Attribute',
    value_name='Value'
)
print("\nReshaped DataFrame:")
print(reshaped)

# 6. Pivoting
pivoted = combined.pivot(
    index='Name',
    columns='Department',
    values='Marks'
)
print("\nPivoted DataFrame:")
print(pivoted)

# 7. Pivot Table
pivot_table = combined.pivot_table(
    index='Department',
    values='Marks',
    aggfunc='mean'
)
print("\nAverage Marks by Department:")
print(pivot_table)

# 8. Stack and Unstack
stacked = pivoted.stack()
print("\nStacked DataFrame:")
print(stacked)

unstacked = stacked.unstack()
print("\nUnstacked DataFrame:")
print(unstacked)
```
