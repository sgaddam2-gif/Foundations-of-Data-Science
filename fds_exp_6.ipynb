Creating Sample DataFrames

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Akhil', 'Ravi', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [4, 5, 6],
    'Name': ['Ram', 'Priya', 'John']
})

print("DataFrame 1:")
print(df1)

print("DataFrame 2:")
print(df2)
```

Concatenation
A. Vertical Concatenation (Stacking Rows)

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Akhil', 'Ravi']
})

df2 = pd.DataFrame({
    'Name': ['Sita', 'Ram']
})

result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)
```

B. Horizontal Concatenation (Stacking Columns)


```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Akhil', 'Ravi']
})

df2 = pd.DataFrame({
    'Marks': [90, 85]
})

result = pd.concat([df1, df2], axis=1)
print(result)
```

Merging



A.inner merge

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Akhil', 'Ravi', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Marks': [85, 90, 75]
})

result = pd.merge(df1, df2, on='ID', how='inner')
print(result)
```

B. Left Merge

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Akhil', 'Ravi', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Marks': [85, 90, 75]
})

result = pd.merge(df1, df2, on='ID', how='left')
print(result)
```

C. Right merge

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Akhil', 'Ravi', 'Sita']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Marks': [85, 90, 75]
})

result = pd.merge(df1, df2, on='ID', how='right')
print(result)
```

Joining

```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Akhil', 'Ravi', 'Sita']
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    'Marks': [85, 90, 75]
}, index=[1, 2, 3])

result = df1.join(df2)
print(result)
```
