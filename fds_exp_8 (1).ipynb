create Data Frames and apply data wrangling methods: Hierarchical Indexing, combining, merging, reshaping ,pivoting, etc.

```python

import pandas as pd

# 1. Create DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'CSE'],
    'Marks': [85, 90, 78]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'City': ['Guntur', 'Vizag', 'Hyderabad']
})

print("Original DataFrame:")
print(df1)

# 2. Hierarchical Indexing
print("\nHierarchical Indexing:")
hierarchy = df1.set_index(['Department', 'Name'])
print(hierarchy)

# 3. Combining DataFrames
print("\nCombining DataFrames:")
df3 = pd.DataFrame({
    'ID': [4, 5],
    'Name': ['Anu', 'Rani'],
    'Department': ['IT', 'CSE'],
    'Marks': [88, 92]
})
combined = pd.concat([df1, df3], ignore_index=True)
print(combined)

# 4. Merging DataFrames
print("\nMerging DataFrames:")
merged = pd.merge(df1, df2, on='ID')
print(merged)

# 5. Reshaping using stack()
print("\nReshaping:")
print(df1.set_index('Name')[['Marks']].stack())

# 6. Pivoting
print("\nPivot Table:")
pivot = df1.pivot_table(
    values='Marks',
    index='Department',
    aggfunc='mean'
)
print(pivot)

# 7. Wide to Long using melt()
print("\nMelt:")
long_data = pd.melt(
    df1,
    id_vars=['Name'],
    value_vars=['Marks'],
    var_name='Subject',
    value_name='Score'
)
print(long_data)

# 8. Joining DataFrames
print("\nJoining DataFrames:")
joined = df1.set_index('ID').join(
    df2.set_index('ID')['City']
)
print(joined)
```
